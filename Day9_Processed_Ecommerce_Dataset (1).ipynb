# Day 9 – Processed E-commerce Dataset

## Pandas Merge, Concatenate, Apply, and DateTime Operations

This notebook combines the provided **Orders**, **Customers**, and **Products** datasets into one clean processed e-commerce dataset.

### Main tasks
- Load three CSV datasets with Pandas
- Combine DataFrames using `concat()`
- Combine related datasets using `merge()`
- Create useful columns using `apply()`
- Convert order dates to DateTime
- Extract year, month, day, and day-of-week information
- Organize the final DataFrame
- Export the processed dataset as a CSV file


## 1. Import Pandas and Load the Three Datasets

```python
import pandas as pd

orders = pd.read_csv("Day9_Orders.csv")
customers = pd.read_csv("Day9_Customers.csv")
products = pd.read_csv("Day9_Products.csv")

print("Orders shape:", orders.shape)
print("Customers shape:", customers.shape)
print("Products shape:", products.shape)
```

## 2. Inspect the Input Datasets

```python
print("Orders:")
display(orders.head())

print("Customers:")
display(customers.head())

print("Products:")
display(products.head())
```

## 3. Combine DataFrames Using `concat()`

`concat()` is demonstrated by splitting the Orders DataFrame into two parts and combining them vertically again. This illustrates row-wise concatenation.


```python
orders_part1 = orders.iloc[:60].copy()
orders_part2 = orders.iloc[60:].copy()

orders_combined = pd.concat(
    [orders_part1, orders_part2],
    ignore_index=True
)

print("First part shape:", orders_part1.shape)
print("Second part shape:", orders_part2.shape)
print("Combined shape:", orders_combined.shape)

display(orders_combined.head())
```

## 4. Merge Orders with Products

The `Product_ID` column is the common key between Orders and Products. A left merge keeps every order and adds the corresponding product information.


```python
orders_products = orders_combined.merge(
    products,
    on="Product_ID",
    how="left",
    validate="many_to_one"
)

print("Shape after merging Orders and Products:", orders_products.shape)
display(orders_products.head())
```

## 5. Merge with Customers

The `Customer_ID` column connects the order information with the Customers DataFrame.


```python
processed = orders_products.merge(
    customers,
    on="Customer_ID",
    how="left",
    validate="many_to_one"
)

print("Shape after merging all three datasets:", processed.shape)
display(processed.head())
```

## 6. Check the Merge Result

```python
print("Missing product names:", processed["Product_Name"].isna().sum())
print("Missing customer names:", processed["Customer_Name"].isna().sum())

print("\nColumns after merging:")
print(processed.columns.tolist())
```

## 7. Convert Order Date to DateTime

The Order_Date column is converted from text to Pandas DateTime so that date components can be extracted easily.


```python
processed["Order_Date"] = pd.to_datetime(processed["Order_Date"])

print("Order_Date data type:", processed["Order_Date"].dtype)
display(processed[["Order_ID", "Order_Date"]].head())
```

## 8. Extract Date Information

Using the Pandas `.dt` accessor, we extract:
- Year
- Month number
- Month name
- Day of the month
- Day of the week


```python
processed["Order_Year"] = processed["Order_Date"].dt.year
processed["Order_Month"] = processed["Order_Date"].dt.month
processed["Order_Month_Name"] = processed["Order_Date"].dt.month_name()
processed["Order_Day"] = processed["Order_Date"].dt.day
processed["Order_Day_Name"] = processed["Order_Date"].dt.day_name()

display(
    processed[
        [
            "Order_ID",
            "Order_Date",
            "Order_Year",
            "Order_Month",
            "Order_Month_Name",
            "Order_Day",
            "Order_Day_Name"
        ]
    ].head()
)
```

## 9. Create Useful Columns Using `apply()`

First, calculate the sales amount from quantity and unit price. Then use `apply()` to classify each order as Low, Medium, or High Value.


```python
processed["Sales_Amount"] = processed.apply(
    lambda row: row["Quantity"] * row["Unit_Price"],
    axis=1
)

processed["Order_Size"] = processed["Sales_Amount"].apply(
    lambda amount: "High Value" if amount >= 5000
    else "Medium Value" if amount >= 2000
    else "Low Value"
)

display(
    processed[
        [
            "Order_ID",
            "Product_Name",
            "Quantity",
            "Unit_Price",
            "Sales_Amount",
            "Order_Size"
        ]
    ].head(10)
)
```

## 10. Create a Clean Final DataFrame

The final dataset is organized so that order, date, customer, product, sales, payment, and status information are easy to understand.


```python
processed = processed[
    [
        "Order_ID",
        "Order_Date",
        "Order_Year",
        "Order_Month",
        "Order_Month_Name",
        "Order_Day",
        "Order_Day_Name",
        "Customer_ID",
        "Customer_Name",
        "City",
        "Region",
        "Membership_Type",
        "Product_ID",
        "Product_Name",
        "Category",
        "Brand",
        "Unit_Price",
        "Quantity",
        "Sales_Amount",
        "Order_Size",
        "Payment_Method",
        "Order_Status"
    ]
]

print("Final shape:", processed.shape)
display(processed.head())
```

## 11. Check the Final Dataset

```python
print("Data types:")
print(processed.dtypes)

print("\nMissing values:")
print(processed.isna().sum())

print("\nDuplicate rows:", processed.duplicated().sum())
```

## 12. Preview Selected Processed Information

```python
display(
    processed[
        [
            "Order_ID",
            "Order_Date",
            "Customer_Name",
            "City",
            "Product_Name",
            "Category",
            "Quantity",
            "Sales_Amount",
            "Payment_Method",
            "Order_Status"
        ]
    ].head(10)
)
```

## 13. Export the Processed Dataset

The final cleaned and enriched DataFrame is exported as `Processed_Ecommerce_Dataset.csv`.


```python
output_file = "Processed_Ecommerce_Dataset.csv"

processed.to_csv(output_file, index=False)

print(f"Processed dataset exported successfully: {output_file}")
```

## 14. Basic Observations

These observations are calculated from the processed dataset.


```python
total_sales = processed["Sales_Amount"].sum()
total_quantity = processed["Quantity"].sum()
top_category = processed.groupby("Category")["Sales_Amount"].sum().idxmax()
top_city = processed.groupby("City")["Sales_Amount"].sum().idxmax()
top_product = processed.groupby("Product_Name")["Sales_Amount"].sum().idxmax()
top_payment = processed.groupby("Payment_Method")["Sales_Amount"].sum().idxmax()

print(f"1. The processed dataset contains {processed.shape[0]} order records and {processed.shape[1]} columns.")
print(f"2. Total sales value calculated from Quantity × Unit_Price is ₹{total_sales:,.2f}.")
print(f"3. Total quantity of products ordered is {total_quantity}.")
print(f"4. The category with the highest total sales is {top_category}.")
print(f"5. The city with the highest total sales is {top_city}.")
print(f"6. The product with the highest total sales is {top_product}.")
print(f"7. The payment method with the highest total sales is {top_payment}.")
```
